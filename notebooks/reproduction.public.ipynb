This notebook reproduces the repaired pipeline with the repository's synthetic fixture. It is not a real-weather or design calculation.

```python
from pathlib import Path
import sys

repo_root = Path.cwd()
if not (repo_root / 'src').exists():
    raise RuntimeError('Open this notebook from the repository root.')
sys.path.insert(0, str(repo_root / 'src'))
```

```python
from residential_cooling_load import parse_building_config, run_weather_series
from residential_cooling_load.model import read_weather_csv

config = parse_building_config(repo_root / 'examples' / 'building.txt')
weather = read_weather_csv(repo_root / 'examples' / 'weather_synthetic.csv')
rows = run_weather_series(config, weather, indoor_temp_c=24.0)
rows
```

```python
peak = max(rows, key=lambda row: row['cooling_load_w'])
assert round(peak['cooling_load_w'], 3) == 3491.187
assert peak['timestamp'] == 'synthetic-summer-noon'
peak
```
